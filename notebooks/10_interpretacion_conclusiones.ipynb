## **Ejercicio 10. Interpretación, limitaciones y conclusiones**

---

## **10.1. Hallazgos en el contexto de participación y consumo de contenido en YouTube**

La participación es **concentrada y de un solo uso**: de 293 videos solo 19 tienen comentarios, y
uno solo (*Qué rico come tu diputado*) concentra el 39.7 % de toda la participación. El 97.3 % de
los autores comentó una sola vez y nunca vuelve a aparecer. Esto es típico del consumo de noticias en
YouTube: la gente reacciona a una pieza puntual (casi siempre denuncias de corrupción, con 61.3 % de
tono negativo) y no sostiene una conversación entre videos. Los pocos autores que sí comentan en más
de un video son justo los que mantienen unida a la red; sin ellos, se cae en pedazos. Además, tener
muchas vistas no garantiza mucha participación (la relación es solo moderada).

---

## **10.2. Limitaciones**

- **Pocos videos tienen comentarios.** Solo 19 de 293 (6.5 %); algunos con millones de vistas no
  tienen ninguno recolectado.
- **La muestra fue dirigida.** Se buscó por temas y canales específicos, así que hay más noticias y
  política de lo que habría en una muestra al azar.
- **Las fechas son relativas** ("hace 2 días"), no exactas, así que un análisis fino de tiempos no es
  posible.
- **Los conteos son de un solo momento.** Un video reciente tiene menos tiempo para acumular vistas
  que uno viejo, así que compararlos por popularidad no es justo.
- **No sabemos quién responde a quién.** `reply_count` solo dice cuántas respuestas hubo, no quién
  las escribió; las conexiones entre autores aquí son solo "comentaron el mismo video", nunca una
  conversación real.
- **Un video domina el conteo.** El más comentado aporta el 40 % del total, así que cualquier
  promedio general está sesgado por él.

---

## **10.3. Descripción, asociación e inferencia**

Lo que sí podemos decir: datos **descriptivos** (cuántos comentaron, qué tan negativo fue el tono,
qué nodos son claves) y **asociaciones** dentro de esta muestra (videos de denuncia → más negativos;
autores que comentan en varios videos → más centrales). Lo que **no** podemos decir es que un tema
"cause" más participación, ni que un canal de gobierno "cause" menos negatividad; podría deberse
igual al tema tratado o al sesgo de cómo se armó la muestra. Y como la muestra cubre solo 19 de 293
videos y fue elegida por búsquedas específicas, **no se puede generalizar** a todo YouTube ni a toda
Guatemala.

---

## **10.4. Conclusiones**

La red es más una **colección de grupos aislados** que una comunidad conectada: 351 nodos, muy poca
densidad, 10 partes sin conexión entre sí, y basta con quitar 17 nodos clave para partirla más.
Louvain confirma lo mismo: encuentra 17 comunidades, pero casi todas son "un video con su gente", no
temas que crucen varios videos.

En cuanto a **sentimiento**, domina el tono negativo (61.3 %), concentrado en videos de denuncia
política; el único video institucional sobre obra pública sale positivo. Negativo no siempre es
rechazo: a veces el enojo es con el hecho denunciado, no con el canal.

En **centralidad**, unos pocos nodos sostienen casi toda la cohesión de la red: un video grande y
siete autores que comentan en más de un lugar.

**En conjunto**: la audiencia reacciona con comentarios sobre todo negativos a piezas puntuales de
denuncia, y participa una sola vez. Estas conclusiones solo aplican a los 19 videos y 406 comentarios
de esta muestra, no a todo el catálogo ni a la audiencia general de YouTube en Guatemala.
